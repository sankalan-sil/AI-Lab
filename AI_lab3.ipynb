Problem 3: Checking Routes Between Cities Using DFS
Problem Statement:
A small road network connects several cities. Write a Python program using DFS that:
1. Checks whether a route exists between two given cities.
2. Finds all separate groups of cities that are connected to each other but not to any
other group (these are called connected components).

```python
#problem 3 : checking routes between cities using DFS

graph = {
    'Delhi': ['Mumbai', 'Jaipur'],
    'Mumbai': ['Delhi', 'Pune'],
    'Jaipur': ['Delhi'],
    'Pune': ['Mumbai'],

    'Kolkata': ['Bhubaneswar'],
    'Bhubaneswar': ['Kolkata'],

    'Chennai': ['Bangalore'],
    'Bangalore': ['Chennai']
}



def route_exists(graph, start, destination):

    visited = set()

    def dfs(city):

        # Destination found
        if city == destination:
            return True

        visited.add(city)

        for neighbour in graph.get(city, []):

            if neighbour not in visited:

                if dfs(neighbour):
                    return True

        return False

    return dfs(start)



def find_connected_components(graph):

    visited = set()
    components = []

    # Include every city appearing in the graph
    cities = set(graph.keys())

    for neighbours in graph.values():
        cities.update(neighbours)

    for city in cities:

        if city not in visited:

            component = []

            def dfs(node):

                visited.add(node)
                component.append(node)

                for neighbour in graph.get(node, []):

                    if neighbour not in visited:
                        dfs(neighbour)

            dfs(city)

            components.append(component)

    return components



start_city = 'Delhi'
destination_city = 'Pune'

if route_exists(graph, start_city, destination_city):
    print(
        f"Route exists between {start_city} "
        f"and {destination_city}."
    )
else:
    print(
        f"No route exists between {start_city} "
        f"and {destination_city}."
    )


components = find_connected_components(graph)

print("\nConnected Components:")

for i, component in enumerate(components, start=1):

    print(
        f"Component {i}: "
        + " -> ".join(sorted(component))
    )
```
